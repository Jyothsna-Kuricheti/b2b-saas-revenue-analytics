```python
# 1. Dataset Creation (Aligned with Power BI Dashboard)
data = {
    'company_id': [101, 102, 103, 104, 105, 106, 107, 108],
    'company_name': [
        'Apex Health Systems', 'SecureNet Shield', 'ShopNova Commerce', 
        'FinEdge Global', 'SwiftFleet Logistics', 'LearnSphere Labs', 
        'CloudWave Tech', 'HealthPlus Clinics'
    ],
    'industry': ['Healthcare', 'Cybersecurity', 'E-commerce', 'Fintech', 'Logistics', 'Edtech', 'Cybersecurity', 'Healthcare'],
    'plan_name': ['Enterprise', 'Enterprise', 'Pro', 'Pro', 'Starter', 'Starter', 'Pro', 'Starter'],
    'arr': [240000, 175000, 48000, 60000, 12000, 15000, 55000, 18000],
    'churn_status': ['Active', 'Active', 'Churned', 'Active', 'Churned', 'Churned', 'Churned', 'Churned'],
    'avg_csat': [4.5, 4.8, 2.0, 3.8, 2.2, 1.5, 2.5, 2.3],
    'daily_active_users': [120, 85, 12, 45, 5, 3, 8, 4],
    'open_support_tickets': [0, 1, 4, 1, 3, 5, 2, 2]
}

df = pd.DataFrame(data)

# 2. Risk Detection Logic (CSAT < 3.0 or High Open Tickets)
at_risk_df = df[(df['avg_csat'] < 3.0) | (df['open_support_tickets'] >= 3)]
print("=== ACCOUNTS AT RETENTION RISK ===")
display(at_risk_df[['company_name', 'plan_name', 'arr', 'avg_csat', 'open_support_tickets']])

# 3. GenAI Retention Copilot Function
def generate_retention_plan(company_row, api_key):
    client = genai.Client(api_key=api_key)
    
    prompt = f"""
    You are an executive B2B SaaS Customer Retention Strategist.
    Analyze the following high-risk client account metrics:
    - Company Name: {company_row['company_name']}
    - Plan Tier: {company_row['plan_name']} (ARR: ${company_row['arr']:,})
    - Churn Status: {company_row['churn_status']}
    - CSAT Score: {company_row['avg_csat']} / 5.0
    - Daily Active Users: {company_row['daily_active_users']}
    - Open Support Tickets: {company_row['open_support_tickets']}

    Provide an executive diagnosis covering:
    1. Root Cause Analysis (RCA): Why this account is burning/disengaged.
    2. Recommended Intervention: Concrete steps for Customer Success team.
    3. Personalized Retention Email Draft: A professional, empathetic email to the client executive offering immediate escalation and product audit.
    """
    
    response = client.models.generate_content(
        model='gemini-3.8-flash',
        contents=prompt
    )
    return response.text

# 4. Run Live AI Diagnostics on Target Account
target_client = at_risk_df.iloc[0]
print(f"\n--- Running AI Copilot for: {target_client['company_name']} ---")

API_KEY = "YOUR_GEMINI_API_KEY_HERE""
output = generate_retention_plan(target_client, api_key=API_KEY)
print(output)
```
